# الواجب الأول: Assignment-1

تطبيق خط أنابيب بيانات RAG أساسي باستخدام تقسيم ثابت الحجم مع تداخل (Fixed-Size Chunking with Overlap) وحساب التشابه باستخدام Cosine Similarity.

````python
import numpy as np
from sentence_transformers import SentenceTransformer

# 1. دالة تحميل الملف النصي المستقل
def load_text_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        return f.read()

# 2. تقسيم النص بحجم نافذة ثابت وتداخل (Fixed-Size Window with Overlap)
def fixed_size_chunking(text, chunk_size=250, overlap_pct=0.15):
    overlap_size = int(chunk_size * overlap_pct)
    chunks = []
    start = 0

    while start < len(text):
        end = start + chunk_size
        chunk = text[start:end]
        chunks.append(chunk)
        start += (chunk_size - overlap_size)
        if chunk_size - overlap_size <= 0:
            break
    return chunks

# 3. دالة حساب تشابه الجيب تماماً (Cosine Similarity)
def cosine_similarity_manual(v1, v2):
    dot_product = np.dot(v1, v2)
    norm_v1 = np.linalg.norm(v1)
    norm_v2 = np.linalg.norm(v2)
    if norm_v1 == 0 or norm_v2 == 0:
        return 0.0
    return dot_product / (norm_v1 * norm_v2)

# تحميل النموذج لتوليد المتجهات الرقمية
print("جاري تحميل نموذج الـ Embeddings...")
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')

# قراءة الملف المستقل وتطبيق التقسيم
raw_text = load_text_file('document_assignment1.txt')
text_chunks = fixed_size_chunking(raw_text, chunk_size=300, overlap_pct=0.15)

# تحويل الفقرات المستخرجة إلى متجهات
chunk_embeddings = model.encode(text_chunks)

print(f"تم تحميل الملف المستقل وقراءته بنجاح! تم تقسيم النص إلى {len(text_chunks)} مقاطع.")
````

### البحث واسترجاع أفضل 3 نتائج في الواجب الأول

````python
# استقبال استعلام البحث من المستخدم
query = "ما هو الذكاء الاصطناعي وتقنية RAG؟"

# تحويل الاستعلام إلى متجه باستخدام نفس النموذج
query_embedding = model.encode(query)

# حساب التشابه بين متجه الاستعلام وجميع المتجهات المخزنة
similarities = []
for idx, chunk_emb in enumerate(chunk_embeddings):
    score = cosine_similarity_manual(query_embedding, chunk_emb)
    similarities.append((score, text_chunks[idx]))

# ترتيب النتائج تنازلياً حسب نسبة التشابه واختيار أفضل 3
similarities.sort(key=lambda x: x[0], reverse=True)
top_k = similarities[:3]

# طباعة النتائج بشكل منظم
print(f"الاستعلام (Question): {query}\n")
print("الأجوبة الأكثر شبهاً وملاءمة (Top 3 Answers):\n")
for i, (score, chunk) in enumerate(top_k, 1):
    print(f"الترتيب {i}: [نسبة التشابه: {score:.4f}]")
    print(f"المحتوى: {chunk.strip()}")
    print("-" * 50)
````

# الواجب الثاني: Assignment-2

## المرحلة الأولى (Phase 1): تحميل الملفات، التقسيم الدلالي/الديناميكي للجمل وحفظها في قاعدة بيانات FAISS.

````python
import faiss
import re
import os

# قائمة الملفات المستقلة
files_list = ['astronomy.txt', 'medicine.txt', 'computers.txt']

all_chunks = []
metadata = [] # لحفظ اسم الملف المرجعي

# دالة تقسيم دلالي ديناميكي بالاعتماد على الجمل
def dynamic_semantic_chunking(text):
    sentences = re.split(r'(?<=[.!?؟])\s+', text.strip())
    return [s for s in sentences if len(s.strip()) > 0]

# معالجة الملفات النصية المستقلة
for file_name in files_list:
    if os.path.exists(file_name):
        content = load_text_file(file_name)
        chunks = dynamic_semantic_chunking(content)
        for chunk in chunks:
            all_chunks.append(chunk)
            metadata.append(file_name)

# توليد متجهات رقمية وحفظها في كشاف FAISS
print("جاري معالجة الملفات المستقلة الثلاثة وتوليد المتجهات...")
embeddings = model.encode(all_chunks)

dimension = embeddings.shape[1]
index = faiss.IndexFlatIP(dimension)

# توحيد المتجهات لحساب Cosine Similarity بشكل صحيح
faiss.normalize_L2(embeddings)
index.add(embeddings)

print(f"تم حفظ {len(all_chunks)} جُمل دلالية من ملفات مستقلة في قاعدة بيانات FAISS بنجاح!")
````

## المرحلة الثانية (Phase 2): البحث الدلالي باستخدام قاعدة بيانات FAISS واسترجاع المرجع (اسم الملف)

ملاحظة: يمكنك تشغيل هذه الخلية بشكل مستقل تماماً للبحث دون الحاجة لإعادة تنفيذ المرحلة الأولى.

````python
# استعلام المستخدم للمرحلة الثانية
user_query_2 = "ما هي استخدامات الأعشاب والإبر الصينية؟"

# تحويل استعلام المستخدم لمتجه رقمي
query_vector = model.encode([user_query_2])
faiss.normalize_L2(query_vector)

# البحث في كشاف FAISS عن أفضل 3 نتائج
k = min(3, len(all_chunks))
scores, indices = index.search(query_vector, k)

# طباعة المخرجات والنتائج والملف المصدر المرجعي لها
print(f"الاستعلام (Question): {user_query_2}\n")
print("الإجابات والمراجع المسترجعة (Top Answers with References):\n")

for i in range(k):
    match_idx = indices[0][i]
    score = scores[0][i]
    source_file = metadata[match_idx]
    answer_chunk = all_chunks[match_idx]

    print(f"الترتيب {i+1}: [نسبة التشابه: {score:.4f}]")
    print(f"اسم الملف المرجعي: {source_file}")
    print(f"الجواب: {answer_chunk.strip()}")
    print("-" * 50)
````
